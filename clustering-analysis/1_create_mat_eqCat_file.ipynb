# Загрузка каталога землетрясений с выбором основного толчка

```python
import pandas as pd
import os

from src.EqCat import EqCat
eqCat = EqCat()
```

```python
# Выбор расположения исходного каталога землетрясений

dir_in = '../data'
file_in = 'NEIC_Global.csv' # 1900–2025
catalogType = 'USGS'

file_path = '../data/All_boundaries'

# Выбор параметров основого толчка

mainshock = {
    'event': 'Tohoku_2011',
    'time': '2011 03 11 05:47:32.8',
    'latitude': 37.52,
    'longitude': 143.05,
    'depth_km': 20,
    'Mw': 9.1,
    'scalar_moment_Nm': 5.31e29,
    'strike_deg': 203,
    'dip_deg': 10,
    'rake_deg': 88
}
```

```python
# Проверка структуры подгруженного каталога замлетрясений

pd.read_csv(f"{dir_in}/{file_in}").info()
```

```python
eqCat.loadEqCat(f"{dir_in}/{file_in}", catalogType, usecols = (0,11,1,2,3,4))
print('total no. of events: ', eqCat.size())
print(sorted(eqCat.data.keys()))
```

```python
eqCat.tieToMainshock(mainshock)
print('no. of events after filtering relative to the mainshock: ', eqCat.size())
saved_file = f"{dir_in}/{os.path.splitext(file_in)[0]}_{mainshock['event']}.mat"
eqCat.saveMatBin(saved_file)
print('Saved file name: ', saved_file)
```
